```python
# Compute per-model lowest NMSE(dB) and overall lowest from the current `metrics` dict
import pandas as pd

epochs = list(range(1, 51))

rows = []
for model, data in metrics.items():
    nmse = data['val_nmse_db']
    min_val = min(nmse)                          # most negative (best) dB
    first_idx = nmse.index(min_val)              # first occurrence
    rows.append({
        "Model": model,
        "Min NMSE (dB)": min_val,
        "Epoch (1-indexed)": epochs[first_idx],
    })

summary_df = pd.DataFrame(rows).sort_values("Min NMSE (dB)").reset_index(drop=True)

# Overall lowest across all models
overall = None
for model, data in metrics.items():
    nmse = data['val_nmse_db']
    min_val = min(nmse)
    idx = nmse.index(min_val)
    if overall is None or min_val < overall["Min NMSE (dB)"]:
        overall = {"Model": model, "Epoch (1-indexed)": epochs[idx], "Min NMSE (dB)": min_val}
overall_df = pd.DataFrame([overall])

# Print as plain pandas DataFrames
print("Per-model Lowest Validation NMSE(dB)")
print(summary_df.to_string(index=False))

print("\nOverall Lowest Validation NMSE(dB)")
print(overall_df.to_string(index=False))
```

```python
# Compute per-model lowest Validation NMSE(dB) and the overall lowest (first occurrence)
import pandas as pd

rows = []
for model, m in metrics.items():
    nmse = m['val_nmse_db']
    min_val = min(nmse)                 # most negative dB = best
    first_idx = nmse.index(min_val)     # first occurrence index
    rows.append({
        "Model": model,
        "Min NMSE (dB)": min_val,
        "Epoch (1-indexed)": first_idx + 1
    })

# Sort by the minimum NMSE(dB) ascending (most negative first)
summary_df = pd.DataFrame(rows).sort_values("Min NMSE (dB)").reset_index(drop=True)

# Overall lowest across all models (first occurrence)
overall = min(
    (
        {
            "Model": model,
            "Epoch (1-indexed)": m['val_nmse_db'].index(min(m['val_nmse_db'])) + 1,
            "Min NMSE (dB)": min(m['val_nmse_db'])
        }
        for model, m in metrics.items()
    ),
    key=lambda x: x["Min NMSE (dB)"]
)
overall_df = pd.DataFrame([overall])

# Print as plain pandas DataFrames
print("Per-model Lowest Validation NMSE(dB)")
print(summary_df.to_string(index=False))

print("\nOverall Lowest Validation NMSE(dB)")
print(overall_df.to_string(index=False))
```
